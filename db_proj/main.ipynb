```python
class t_process(db.Model):
    __table_args__ = {'extend_existing':True}
    process_id = db.Column(db.Integer,primary_key=True,autoincrement=True)
    name = db.Column(db.String(255))
    course_id = db.Column(db.Integer)
    class_id = db.Column(db.Integer)
    time = db.Column(db.Date)

    def to_dict(self):
        return {
            "process_id":self.process_id,
            "name":self.name,
            "course_id":self.course_id,
            "class_id":self.class_id,
            "time":self.time
        }
```

```python
with app.app_context():
    class_ls = Dto.getDTOls(t_class)
    for class_i in class_ls:
        course = class_i.course_id
        jie = random.randint(1,8)
        startTime = date(2025,2,18)
        for pro_i in range(random.randint(5,10)):
            proc = t_process()
            proc.name =  "第"+str(pro_i+1)+"周第"+str(jie)+"节"
            proc.course_id = course
            proc.time = startTime
            proc.class_id = class_i.class_id
            startTime += timedelta(weeks=1)
            print(proc.to_dict())
            db.session.add(proc)
    db.session.commit()
```

# 为学生分配对应的任务
- 题目id为-1

任务为主体
1. 获得任务列表
2. 找到任务对应的班级 -> 1个
3. 班级对应的学生 -> 列表
4. 对每个学生设置状态

```python
with app.app_context():
    # 获得任务列表
    exer_ls = Dto.getDTOls(t_exer)
    # 循环对于每一个任务
    for exer in exer_ls:
        # 找到任务对应的班级 -> 1个
        now_class_id = exer.class_id
        if now_class_id == -1:
            continue
        # 班级对应学生 -> list
        stuid_ls =  t_user_course_class.get_stuid_ls_by_classid(now_class_id)
        # debug
        # print("classid = {}".format(exer.class_id))
        # print(stuid_ls)
        for stuid in stuid_ls:
            rec = t_stu_prob_exer(
                stu_id = stuid[0],
                prob_id = -1,
                exer_id = exer.exer_id,
                score = random.randint(60,100),
                is_finish = MyUtils.getRandomBoolean(),
                is_check = MyUtils.getRandomBoolean()
            )
            db.session.add(rec)
    db.session.commit()
    
    
```

# 学生 题目 任务 记录
- 首先找到学生和任务的对应组
- 根据任务，找到对应的题目列表
    - 令 stuid = -1 得到对应任务的所有题目
- 循环每一个pair组
- 内层循环找到的题目列表
    - 根据题目id得到type 生成对应的submit

- 外层循环每一个学生的每一次任务
    - 内层循环每一个任务的若干道题目

```python
with app.app_context():
    stu_exer_pair_list = t_stu_prob_exer.stu_exer_pair_ls()
    for stu_exer_pair in stu_exer_pair_list:
        exer_id = stu_exer_pair[0]
        stu_id = stu_exer_pair[1]
        prob_id_list = t_stu_prob_exer.get_prob_id_list_by_exer_id(exer_id)
        id = 1
        for prob_id in prob_id_list:
            rec = t_stu_prob_exer(
                stu_id = stu_id,
                prob_id = prob_id,
                exer_id = exer_id,
                score = random.randint(3,5),
                comment = t_stu_prob_exer.get_random_comment(),
                submit = t_prob.gen_answer_by_type(t_prob.get_prob_by_id(prob_id).type),
                is_finish=MyUtils.getRandomBoolean(),
                is_check = MyUtils.getRandomBoolean(),
                idx = id
            )
            id += 1
            # db.session.add(rec)
            print(rec.to_dict())
    # db.session.commit()
```

```python
with app.app_context():
    stu_exer_pair_list = t_stu_prob_exer.stu_exer_pair_ls()
    for stu_exer_pair in stu_exer_pair_list:
        exer_id = stu_exer_pair[0]
        stu_id = stu_exer_pair[1]
        prob_id_list = t_stu_prob_exer.get_prob_id_list_by_exer_id(exer_id)
        for prob_id in prob_id_list:
            if MyUtils.random_by_probability(0.2):
                rec = t_wrong(
                    stu_id = stu_id,
                    prob_id = prob_id,
                    exer_id = exer_id
                )
                db.session.add(rec)
            if MyUtils.random_by_probability(0.15) and len(t_collect.query.filter_by(stu_id=stu_id,prob_id=prob_id).all()) == 0:
                rec = t_collect(
                    stu_id = stu_id,
                    prob_id = prob_id
                )
                db.session.add(rec)
    db.session.commit()
```
